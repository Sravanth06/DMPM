3. Correlation co-efficient relation and co-variance analysis.

```python
import pandas as pd
import numpy as np

# Sample data
data = {
    "X": [10, 20, 30, 40, 50],
    "Y": [15, 25, 35, 45, 60]
}

df = pd.DataFrame(data)

# -----------------------------
# 1. Covariance
# -----------------------------
covariance = df["X"].cov(df["Y"])
print("Covariance:", covariance)

# -----------------------------
# 2. Correlation Coefficient
# -----------------------------
correlation = df["X"].corr(df["Y"])
print("Correlation Coefficient:", correlation)

# -----------------------------
# 3. Correlation Matrix
# -----------------------------
correlation_matrix = df.corr()
print("\nCorrelation Matrix:")
print(correlation_matrix)

# -----------------------------
# 4. Covariance Matrix
# -----------------------------
covariance_matrix = df.cov()
print("\nCovariance Matrix:")
print(covariance_matrix)

# -----------------------------
# 5. Relationship interpretation
# -----------------------------
if correlation > 0:
    print("\nThere is a positive relationship between X and Y.")
elif correlation < 0:
    print("\nThere is a negative relationship between X and Y.")
else:
    print("\nThere is no linear relationship between X and Y.")

if abs(correlation) >= 0.7:
    print("The relationship is strong.")
elif abs(correlation) >= 0.4:
    print("The relationship is moderate.")
else:
    print("The relationship is weak.")
```
